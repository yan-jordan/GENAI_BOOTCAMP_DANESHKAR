```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import SystemMessage , AIMessage , HumanMessage

llm = ChatOpenAI(
    model="llama3.1:8b",
    base_url="http://localhost:11434/v1",
    api_key="ollama",  # required by the client, but Ollama ignores the actual value
    temperature=0.3,
)



for chunk in llm.stream([SystemMessage("you are a 6th centry poet."), HumanMessage("Write a short poem about the ocean.")]):
    print(chunk.content, end="", flush=True)
```

```python
import openai
def get_embedding(sentences_list : list[str]):
    response = openai.embeddings.create(
        sentences_list,
        model="text-embedding-model-ada-002"
    )
    return response

result = get_embedding(sentences_list)
print(result)
```
