# ENGLISH DOCUMENTS

```python
import os
from dotenv import load_dotenv

load_dotenv()

api_key = os.getenv("OPENROUTER_API_KEY")
```

```python
# Local LLM model ---> ollama ---> llama3.1:8b
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    base_url="http://localhost:11434/v1",
    model_name="llama3.1:8b",
    api_key="ollama",
    streaming=True,
)

for chunk in llm.stream("please introduce your self in few words!"):
    print(chunk.content, end="", flush=True)
```

```python
# open ai embedding model for english documents
from langchain_openai import OpenAIEmbeddings
open_ai_embedding_model = OpenAIEmbeddings(
    model= "text-embedding-3-small",
    api_key= api_key,
    base_url= "https://openrouter.ai/api/v1"
)
```

```python
from langchain_community.document_loaders import TextLoader , DirectoryLoader

loader = TextLoader("./data/messi/messi.txt")
english_documents = loader.load()

print(len(english_documents))
print(english_documents)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

english_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 1000,
    chunk_overlap = 100
)

english_chunks = english_splitter.split_documents(english_documents)
print(english_chunks)
```

```python
print(len(english_chunks))
print(type(english_chunks[2]))
```

```python
from langchain_community.vectorstores import FAISS

english_db = FAISS.from_documents(
    documents= english_chunks,
    embedding= open_ai_embedding_model   
)

print(english_db)
print(english_db.index)
print(english_db.index.ntotal)
```

```python
english_db.save_local("my_index" , "my_docs")
```

```python
english_db.similarity_search("When did Messi win his record eighth Ballon d'Or?" , filter= { 'source': './data/messi/messi.txt' } ,  k = 1)
```

# PERSIAN DOCUMENTS

```python
# Local persian fine tune embedding model from hugging face
from langchain_huggingface.embeddings import HuggingFaceEmbeddings

model_name = "HooshvareLab/bert-base-parsbert-uncased"
model_kwargs = {'device': 'cpu'}
encode_kwargs = {'normalize_embeddings': False}
hf_embedding = HuggingFaceEmbeddings(
model_name=model_name,
model_kwargs=model_kwargs,
encode_kwargs=encode_kwargs
)
```

```python
from langchain_community.document_loaders import TextLoader , DirectoryLoader
loader = TextLoader("./data/messi/messi_persian.txt")
persian_documents = loader.load()

print(len(persian_documents))
print(persian_documents)
```

```python
# persian / arabic content needs to be normalize
from hazm import Normalizer
n = Normalizer(persian_numbers=False)  # keep digits as-is if you need them searchable
for persian_document in persian_documents:
    persian_document.page_content = n.normalize(persian_document.page_content)
```

```python
persian_splitter = RecursiveCharacterTextSplitter(
    separators=[
        "\n\n",  
        "\n",     
        "؟ ",     
        "! ",     
        ". ",     
        "، ",     
        " ",      
        "",       
    ],
    chunk_size = 1000,
    chunk_overlap = 80,
    keep_separator= True
)

persian_chunks = persian_splitter.split_documents(persian_documents)

print(len(persian_chunks))
print(persian_chunks[3])
```

```python
# time to create a data base and put our chunks on it
from langchain_community.vectorstores import FAISS

persian_db = FAISS.from_documents(
    documents= persian_chunks,
    embedding= hf_embedding,
)

print(persian_db.index.ntotal)
```

```python
query = n.normalize("مسی هشتمین توپ طلای رکوردشکن خود را چه زمانی برد")

response = persian_db.similarity_search(
    query=query
) 

print(response[0].page_content)
```
